```python
import h5py, glob
import sys
sys.path.append('/suphys/aani0116/Work/minicluster_tidal/my_modules')
from mc_declare import *

import matplotlib.pyplot as plt
files = glob.glob('/suphys/aani0116/Work/minicluster_tidal/hpc/tidal_timestep/*.h5')
for i,j in enumerate(files):
    print(i, files[i])
```

```python
with h5py.File(files[0], 'r') as f:
    mcp_xv     = f['mcp_xv'][:]
    bound      = f['bound'][:]
    E_p        = f['E_p'][:]
    mc_xv      = f['mc_xv'][:]
    snap_times = f['snap_times'][:]
    
    mcp_mass        = f.attrs['mcp_mass']
    Nbody           = f.attrs['Nbody']
    simulation_time = f.attrs['simulation_time']
    tupd            = f.attrs['tupd']
    mc_mass         = f.attrs['mc_mass']
    mc_delta        = f.attrs['mc_delta']
    mc_radius       = f.attrs['mc_radius']
    mc_radius_char  = f.attrs['mc_radius_char']
    mc_density_char = f.attrs['mc_density_char']
    mc_concentration= f.attrs['mc_concentration']
fig, axes = plt.subplots(3, 4, figsize=(14, 12))
for i, ax in enumerate(axes.flat):
    if i < 10:
        ax.scatter(mcp_xv[i, :, 0], mcp_xv[i, :, 2], s=0.5)
        ax.set_title(f't = {snap_times[i]:.4f} Gyr')
        ax.set_xlabel('x (kpc)')
        ax.set_ylabel('z (kpc)')
    elif i == 10:
        ax.plot(mc_xv[:, 0], mc_xv[:, 2])

        ax.set_title('MC centre')
        ax.set_xlabel('x (kpc)')
        ax.set_ylabel('z (kpc)')
    else:
        ax.set_visible(False)
plt.tight_layout()
print(tupd,simulation_time)
```
